**Reading Report**:
1. alagar. The Rise of Ethical Data Science: the Moral Landscape. IABAC, Oct. 3, 2023. Accessed 29 Oct. 2025. https://iabac.org/blog/the-rise-of-ethical-data-science-the-moral-landscape

Summary:
This article explains why ethics must be built into each stage of data work. It outlines core principles, like fairness, transparency, accountability, privacy, and walks through common ethical problems like biased training data, “black box” models, and weak consent practices. It also suggests practical approaches, for example, ethical guidelines, impact assessments, diverse teams, documentation, and better organization.

what stood out:
ethics isn’t just abstract, it needs concrete processes

It reinforced that transparency and accountability are operational tasks. That made me think more about who signs off on ethical review.

I agree with the emphasis on embedding ethics into workflows, the piece is further proof that organizations must require and resource ethics work, not treat it as optional.


2. Hand, David J. “Aspects of Data Ethics in a Changing World.” PubMed Central (PMC), 2018. Accessed 29 Oct. 2025. https://pmc.ncbi.nlm.nih.gov/articles/PMC6154451/


Summary:
Hand’s article surveys fundamental ethical issues surrounding modern data practices. the nature and provenance of data, personal data and consent, ownership and stewardship, trustworthiness and data quality, privacy and confidentiality, and the social consequences of data-driven decisions. It frames ethics as a set of trade-offs and argues for careful, context-sensitive judgment rather than one-size-fits-all rules

what stood out:
Hand highlights trade-offs and the limits of technical fixes (e.g., fairness metrics can’t fully replace human judgment). That shows that ethics often requires policy and social judgment, not only algorithms

The discussion that poor or biased data produces ethical problems downstream, so stewardship matters early.

I agree with the article’s caution: ethics requires context, transparency about limitations, and multiple views.
